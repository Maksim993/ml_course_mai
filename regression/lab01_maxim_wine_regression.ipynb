# Лабораторная работа №1 — прогноз качества вина

**Курс:** Машинное обучение  
**Набор данных:** UCI Wine Quality (red wine)

В работе решается задача **регрессии**: по 11 физико-химическим характеристикам вина необходимо предсказать экспертную оценку `quality`.

Порядок эксперимента: загрузка и проверка данных → корректное train/test-разбиение → EDA на train → baseline → Linear Regression → Random Forest → Gradient Boosting → grouped cross-validation → финальная оценка на test → анализ важности признаков и ошибок.

Основная метрика — **RMSE**, дополнительная — **MAE**. Для обученных моделей ориентир лабораторной: `RMSE < 0.70`.

В наборе есть повторяющиеся комбинации признаков, поэтому одинаковые объекты не должны одновременно попадать в train и test.


```python
from pathlib import Path
import hashlib
import json
import platform
import time
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
from sklearn.model_selection import GroupShuffleSplit, GroupKFold, GridSearchCV, cross_validate
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from IPython.display import display, Markdown
SEED = 42
pd.set_option('display.max_columns', 20)
pd.set_option('display.precision', 4)
plt.rcParams.update({'figure.dpi':115, 'axes.grid':True, 'grid.alpha':.2, 'font.size':10, 'axes.spines.top':False, 'axes.spines.right':False})

def save_figure(name):
    """Показать график в ноутбуке без записи отдельных файлов."""
    plt.tight_layout()
    plt.show()
    plt.close()

%matplotlib inline
```

## 1. Подготовка данных и разбиение

CSV читается с разделителем `;`. Проверяем названия столбцов, пропуски и бесконечные значения.

`quality` используется только как целевая переменная. Группы строятся **по 11 признакам X**, без использования целевого ответа.

20% данных откладываются в test. `random_state = 42` фиксирует воспроизводимость. `GroupShuffleSplit` нужен, чтобы одинаковые наборы признаков не оказались одновременно в train и test.


```python
path = Path('winequality-red.csv')
df = pd.read_csv(path, sep=';')
expected = ['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar', 'chlorides', 'free sulfur dioxide', 'total sulfur dioxide', 'density', 'pH', 'sulphates', 'alcohol', 'quality']
if list(df.columns) != expected or df.isna().any().any():
    raise ValueError('Неверная структура датасета или пропуски: требуется проверка данных.')
if not np.isfinite(df.to_numpy()).all():
    raise ValueError('Обнаружены бесконечные значения.')
X, y = (df.drop(columns='quality'), df['quality'])
groups = pd.util.hash_pandas_object(X, index=False).astype(str)
train, test = next(GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=SEED).split(X, groups=groups))
assert set(groups.iloc[train]).isdisjoint(set(groups.iloc[test]))
split = pd.DataFrame({'row_id': df.index, 'subset': 'train', 'group': groups})
split.loc[test, 'subset'] = 'test'
meta = {'dataset': 'UCI Wine Quality: red wine', 'sha256': hashlib.sha256(path.read_bytes()).hexdigest(), 'seed': SEED, 'n_rows': len(df), 'n_features': X.shape[1], 'n_train': len(train), 'n_test': len(test), 'unique_X': int(groups.nunique()), 'exact_duplicate_rows': int(df.duplicated().sum()), 'python': platform.python_version()}
print(json.dumps(meta, ensure_ascii=False, indent=2))
print('Пересечение групп train/test: 0. Пропусков: 0.')
data_split = {'df': df, 'X_train': X.iloc[train], 'X_test': X.iloc[test], 'y_train': y.iloc[train], 'y_test': y.iloc[test], 'groups': groups.iloc[train], 'meta': meta}

display(data_split['X_train'].head())
display(pd.Series({
    'train_rows': len(data_split['X_train']), 'test_rows': len(data_split['X_test']),
    'train_fraction': len(data_split['X_train']) / len(data_split['df']),
    'test_fraction': len(data_split['X_test']) / len(data_split['df'])
}, name='split'))
```

## 2. Первичный анализ обучающей выборки

EDA проводится только на train. Рассматриваем описательные статистики, распределение `quality`, корреляции признаков с целью, полную корреляционную матрицу и несколько заметных зависимостей.

Корреляция показывает статистическую связь, а не причинный эффект.


```python
X, y = (data_split['X_train'], data_split['y_train'])
train = X.assign(quality=y)
summary_stats = train.describe().T
corr = train.corr()
target_correlations = corr['quality'].drop('quality').sort_values()
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
y.value_counts().sort_index().plot.bar(ax=axs[0], color='#3c7ca6', rot=0)
axs[0].set(title='Распределение целевой переменной на train', xlabel='Оценка качества', ylabel='Число объектов')
target_correlations.plot.barh(ax=axs[1], color=['#be554d' if v < 0 else '#3c7ca6' for v in target_correlations])
axs[1].set(title='Связь признаков с quality на train', xlabel='Корреляция')
save_figure('01_target_correlations.png')
fig, ax = plt.subplots(figsize=(11, 8))
im = ax.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1)
ax.set_xticks(range(len(corr)), corr.columns, rotation=70, ha='right')
ax.set_yticks(range(len(corr)), corr.columns)
for i in range(len(corr)):
    for j in range(len(corr)):
        ax.text(j, i, f'{corr.iloc[i, j]:.2f}', ha='center', va='center', fontsize=7, color='white' if abs(corr.iloc[i, j]) > 0.65 else 'black')
fig.colorbar(im, ax=ax, shrink=0.8)
ax.set_title('Матрица корреляций обучающей выборки')
save_figure('02_correlation_matrix.png')
fig, axs = plt.subplots(1, 3, figsize=(13, 4))
for ax, col in zip(axs, ['alcohol', 'volatile acidity', 'sulphates']):
    ax.scatter(X[col], y, s=13, alpha=0.18, color='#3c7ca6')
    ax.set(xlabel=col, ylabel='quality')
save_figure('03_relationships.png')
counts = y.value_counts().sort_index()
positive, negative = (target_correlations.idxmax(), target_correlations.idxmin())
text = f'На train {len(X)} объектов, 11 числовых признаков. Пропусков нет. Оценки 5 и 6 составляют {counts.reindex([5, 6], fill_value=0).sum() / len(y):.1%} train; поэтому низкие и высокие оценки труднее выучить. Сильнейшая положительная линейная связь с quality у {positive} (r={target_correlations[positive]:.3f}), отрицательная — у {negative} (r={target_correlations[negative]:.3f}). Это связи в наблюдениях, а не причинные эффекты. Алкоголь и летучая кислотность могут помогать прогнозу, но одной корреляции недостаточно для точного предсказания. Состав вина связан между собой: коррелирующие признаки могут перераспределять веса и важности. Сохраняем все 11 признаков. Редкие значения не удаляем как выбросы: это часть задачи.'
print(text)
description, eda_text = summary_stats, text

display(description)
display(data_split['y_train'].value_counts().sort_index().rename('train_count').to_frame())
display(Markdown('**Вывод по EDA.** ' + eda_text))
```

## 3. Метрики и baseline

Используем MAE и RMSE. MAE показывает среднюю абсолютную ошибку, а RMSE сильнее штрафует крупные промахи и является основной метрикой.

Baseline — `DummyRegressor(strategy='mean')`: модель всегда выдаёт среднее значение `quality` по train. Все полноценные модели должны быть лучше этого ориентира.


```python
baseline_mean = float(data_split['y_train'].mean())
print(f'Baseline предсказывает {baseline_mean:.6f} для каждого объекта.')
print('На примере [0, 2]: MAE =', np.mean(np.abs([0, 2])),
      ', RMSE =', np.sqrt(np.mean(np.square([0, 2]))))
```

## 4. Cross-validation и обучение моделей

Все модели сравниваются на одинаковых 5 grouped folds внутри train. Так test остаётся нетронутым до финала, а одинаковые объекты не пересекают границу train/validation внутри фолдов.

Обучаются Baseline, Linear Regression со `StandardScaler`, Random Forest и Gradient Boosting. Для двух ансамблей параметры подбираются через `GridSearchCV`. Победитель определяется по минимальному среднему CV RMSE **до просмотра test**.


```python
X, y, groups = (data_split['X_train'], data_split['y_train'], data_split['groups'])
folds = list(GroupKFold(n_splits=5, shuffle=True, random_state=SEED).split(X, y, groups))
for tr, va in folds:
    assert set(groups.iloc[tr]).isdisjoint(set(groups.iloc[va]))
cv_fold_id = pd.Series(-1, index=X.index, name='validation_fold')
for k, (_, va) in enumerate(folds):
    cv_fold_id.iloc[va] = k
scoring = {'RMSE': 'neg_root_mean_squared_error', 'MAE': 'neg_mean_absolute_error'}
models = {'Baseline': DummyRegressor(strategy='mean'), 'Linear Regression': make_pipeline(StandardScaler(), LinearRegression())}
rows = []
for name, model in models.items():
    scores = cross_validate(model, X, y, cv=folds, scoring=scoring, n_jobs=1)
    rows.append({'model': name, 'CV_RMSE': -scores['test_RMSE'].mean(), 'CV_RMSE_std': scores['test_RMSE'].std(), 'CV_MAE': -scores['test_MAE'].mean()})
    model.fit(X, y)
specs = [('Random Forest', RandomForestRegressor(n_estimators=400, random_state=SEED, n_jobs=1), {'max_features': [0.8, 1.0], 'min_samples_leaf': [1, 2]}), ('Gradient Boosting', GradientBoostingRegressor(random_state=SEED, learning_rate=0.05, loss='squared_error'), {'n_estimators': [300, 600], 'max_depth': [2, 3]})]
best_params = {}
for name, model, params in specs:
    print(f'Подбор {name}: 4 кандидата × 5 фолдов...', flush=True)
    search = GridSearchCV(model, params, cv=folds, scoring=scoring, refit='RMSE', n_jobs=1, error_score='raise', return_train_score=True)
    start = time.perf_counter()
    search.fit(X, y)
    print(f'{name}: CV RMSE={-search.best_score_:.4f}; {time.perf_counter() - start:.1f} с', flush=True)
    idx = search.best_index_
    rows.append({'model': name, 'CV_RMSE': -search.best_score_, 'CV_RMSE_std': search.cv_results_['std_test_RMSE'][idx], 'CV_MAE': -search.cv_results_['mean_test_MAE'][idx]})
    models[name] = search.best_estimator_
    best_params[name] = search.best_params_
cv_results = pd.DataFrame(rows).set_index('model')
winner = cv_results['CV_RMSE'].idxmin()
print('Выбрана до оценки на test:', winner)

display(cv_results)
import json
display({'winner': winner, 'best_params': best_params})
display(Markdown(f'**Модель зафиксирована до test:** {winner}.'))
```

## 5. Финальная проверка на test

После cross-validation параметры считаются зафиксированными. Только теперь рассчитываются прогнозы на test. Для каждой модели выводятся RMSE, MAE, train RMSE и CV-метрики.

Дополнительно автоматически проверяется условие `RMSE < 0.70` для Linear Regression, Random Forest и Gradient Boosting. Baseline в это условие не включается.


```python
rows, predictions = ([], {})
y = data_split['y_test']
for name, model in models.items():
    pred = model.predict(data_split['X_test'])
    predictions[name] = pred
    rows.append({'model': name, 'RMSE': root_mean_squared_error(y, pred), 'MAE': mean_absolute_error(y, pred), 'train_RMSE': root_mean_squared_error(data_split['y_train'], model.predict(data_split['X_train']))})
metrics = pd.DataFrame(rows).set_index('model').join(cv_results)
prediction_table = pd.DataFrame(predictions, index=y.index).rename_axis('row_id')
prediction_table.insert(0, 'quality', y)
fig, ax = plt.subplots(figsize=(8, 4))
metrics[['RMSE', 'MAE']].plot.bar(ax=ax, color=['#3c7ca6', '#e8a44f'], rot=15)
ax.axhline(0.7, color='#a43a3a', linestyle='--', label='RMSE: порог 0.70')
ax.set(ylabel='Ошибка в баллах quality', title='Сравнение моделей на отложенном test')
ax.legend(loc='upper right')
save_figure('04_metrics.png')
print('Победитель по CV:', winner)
print(f"Его test RMSE = {metrics.loc[winner, 'RMSE']:.4f}; MAE = {metrics.loc[winner, 'MAE']:.4f}. Порог 0.70: " + ('пройден' if metrics.loc[winner, 'RMSE'] < 0.7 else 'не пройден'))

display(metrics)
display(Markdown(
    f'**Проверка локального порога для выбранной модели:** '
    f'RMSE={metrics.loc[winner, "RMSE"]:.4f}; '
    + ('порог < 0.70 пройден.' if metrics.loc[winner, 'RMSE'] < .70 else 'порог < 0.70 не пройден.')
))
```

## 6. Интерпретация моделей

Для Linear Regression анализируются стандартизованные коэффициенты. Для Random Forest и Gradient Boosting — `feature_importances_`.

Эти значения помогают интерпретировать модель, но не доказывают причинность.


```python
cols = data_split['X_train'].columns
linear_coefficients = pd.Series(models['Linear Regression'].named_steps['linearregression'].coef_, index=cols, name='standardized_coefficient')
feature_importance = pd.DataFrame({name: models[name].feature_importances_ for name in ['Random Forest', 'Gradient Boosting']}, index=cols)
fig, axs = plt.subplots(1, 3, figsize=(16, 5))
linear_coefficients.sort_values().plot.barh(ax=axs[0], color='#3c7ca6')
axs[0].axvline(0, color='black', linewidth=0.7)
axs[0].set(title='Linear Regression: коэффициенты', xlabel='Δ quality при +1 σ признака')
for ax, name in zip(axs[1:], feature_importance.columns):
    feature_importance[name].sort_values().plot.barh(ax=ax, color='#509b83')
    ax.set(title=name, xlabel='Уменьшение impurity (MDI)')
save_figure('05_importance.png')
top = {'Linear Regression': list(linear_coefficients.abs().nlargest(3).index), **{name: list(feature_importance[name].nlargest(3).index) for name in feature_importance.columns}}
print('Топ-3 признака:', json.dumps(top, ensure_ascii=False))
print('Коэффициенты относятся к стандартизованным признакам; это условные связи при фиксированных остальных X. Для обычной линейной регрессии масштабирование почти не меняет предсказания, но позволяет сравнивать коэффициенты. Лес и бустинг не требуют стандартизации. MDI показывает вклад разбиений в уменьшение ошибки на train, не направление эффекта. MDI может предпочитать признаки с большим числом порогов; коррелирующие X делят важность. Причинный вывод невозможен: это наблюдательные данные без контролируемого вмешательства.')
coefficients, importances = linear_coefficients, feature_importance

display(coefficients.reindex(coefficients.abs().sort_values(ascending=False).index).to_frame())
display(importances.sort_values('Random Forest', ascending=False))
corr = data_split['X_train'].corrwith(data_split['y_train'])
corr_top = set(corr.abs().nlargest(3).index)
lines = []
for name, names in top.items():
    overlap = sorted(set(names) & corr_top)
    lines.append(f'**{name}:** топ-3 — {", ".join(names)}. '
                 f'Совпадение с топ-3 EDA по |корреляции|: {", ".join(overlap) or "нет"}.')
common = sorted(set(top['Random Forest']) & set(top['Gradient Boosting']))
lines.append('**Общие лидеры двух ансамблей:** ' + (', '.join(common) or 'нет') + '.')
lines.append('Различия ожидаемы: корреляция измеряет отдельную линейную связь, коэффициент — '
             'условную линейную связь, а MDI учитывает пороги и взаимодействия. '
             'Взаимосвязанные признаки могут делить важность, поэтому рейтинги не обязаны совпадать.')
display(Markdown('\n\n'.join(lines)))
```

### Почему важность не равна причинному эффекту

Данные наблюдательные, признаки взаимосвязаны, а часть свойств вина может отсутствовать в таблице. Поэтому коэффициенты, корреляции и `feature_importances_` описывают закономерности модели, а не доказанный эффект изменения конкретного вещества.


## 7. Анализ ошибок лучшей модели по CV

Для выбранной модели рассчитывается `residual = quality - prediction`, строятся распределение ошибок, график истинных и предсказанных значений, метрики по каждому значению `quality` и таблица крупнейших ошибок.

Для трёх самых больших ошибок дополнительно ищутся ближайшие объекты train в стандартизованном пространстве признаков. Этот анализ выполняется уже после фиксации модели и не используется для повторной настройки.


```python
y = data_split['y_test']
pred = predictions[winner]
error_table = data_split['X_test'].copy()
error_table['quality'] = y
error_table['prediction'] = pred
error_table['residual'] = y - pred
error_table['absolute_error'] = abs(y - pred)
error_table.index.name = 'row_id'
largest_errors = error_table.sort_values('absolute_error', ascending=False).head(10)
grouped = []
for q, group in error_table.groupby('quality'):
    grouped.append({'quality': int(q), 'n_test': len(group), 'n_train': int((data_split['y_train'] == q).sum()), 'mean_prediction': group['prediction'].mean(), 'MAE': group['absolute_error'].mean(), 'RMSE': root_mean_squared_error(group['quality'], group['prediction'])})
quality_error_table = pd.DataFrame(grouped).set_index('quality')
fig, axs = plt.subplots(1, 3, figsize=(15, 4))
axs[0].hist(error_table['residual'], bins=20, color='#3c7ca6', edgecolor='white')
axs[0].axvline(0, color='black', linestyle='--')
axs[0].set(title='Ошибки выбранной модели', xlabel='quality − prediction', ylabel='Число объектов')
axs[1].scatter(y, pred, alpha=0.45, s=18, color='#3c7ca6')
lim = [min(y.min(), pred.min()) - 0.2, max(y.max(), pred.max()) + 0.2]
axs[1].plot(lim, lim, '--', color='#a43a3a')
axs[1].set(xlabel='Истинное quality', ylabel='Предсказание', title=winner)
quality_error_table[['RMSE', 'MAE']].plot.bar(ax=axs[2], rot=0, color=['#3c7ca6', '#e8a44f'])
axs[2].set(title='Ошибка по истинной оценке', ylabel='Ошибка в баллах quality')
save_figure('06_errors.png')
scaler = StandardScaler().fit(data_split['X_train'])
scaled_train = scaler.transform(data_split['X_train'])
error_comments = []
for row_id, row in largest_errors.head(3).iterrows():
    sample = data_split['X_test'].loc[[row_id]]
    d = np.linalg.norm(scaled_train - scaler.transform(sample), axis=1)
    near = np.argsort(d)[:5]
    near_y = data_split['y_train'].iloc[near].tolist()
    z = pd.Series(scaler.transform(sample)[0], index=sample.columns)
    unusual = ', '.join((f'{k}: {v:+.1f}σ' for k, v in z.reindex(z.abs().nlargest(2).index).items()))
    note = f"Строка {row_id}: истинное quality={row['quality']:.0f}, прогноз={row['prediction']:.2f}, абсолютная ошибка={row['absolute_error']:.2f}. Оценки 5 ближайших train-объектов: {near_y}; сильнейшие отклонения от train-средних: {unusual}. Возможное объяснение — похожие химические характеристики соответствуют другим экспертным оценкам; также могут влиять неучтённые свойства вина и шум оценивания. Это гипотеза, а не доказанная причина ошибки."
    error_comments.append(note)
    print(note)
print('Редкие оценки слабее представлены в обучении, поэтому при квадратичной функции потерь прогнозы тяготеют к более частым средним значениям. Для малых групп метрики нестабильны; смотрите n_test и n_train. Этот анализ выполнен после фиксации модели и не используется для её донастройки.')
worst_errors, errors_by_quality, error_notes = largest_errors, quality_error_table, error_comments

display(errors_by_quality)
display(worst_errors)
for note in error_notes:
    display(Markdown(note))
display(Markdown(
    'Редкие крайние оценки представлены слабо, а минимум квадратичной ошибки тяготеет '
    'к условному среднему. Поэтому модель может завышать низкие и занижать высокие '
    'оценки. Это следует проверять по таблице, а не предполагать одинаковую точность для всех групп. '
    'После этого анализа параметры и состав данных не меняются.'
))
```

## 8. Итог эксперимента

Финальный вывод включает сравнение с baseline, качество модели-победителя по cross-validation, поведение на test и ограничения эксперимента.
